```python
%pip install nltk xgboost
import numpy as np 
import nltk
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split 
from xgboost import XGBClassifier 
from sklearn.metrics import classification_report, f1_score
```

```python
dataset_files = [
    'aggression_parsed_dataset.csv',
    'attack_parsed_dataset.csv',
    'kaggle_parsed_dataset.csv',
    'toxicity_parsed_dataset.csv',
    'twitter_parsed_dataset.csv',
    'twitter_racism_parsed_dataset.csv',
    'twitter_sexism_parsed_dataset.csv',
    'youtube_parsed_dataset.csv'
]
combined_dfs = []
for file in dataset_files:
    df = pd.read_csv(file)
    if 'Text' in df.columns and 'oh_label' in df.columns:
        sub_df = df[['Text', 'oh_label']].dropna()
        combined_dfs.append(sub_df)
```

```python
combined_df = pd.concat(combined_dfs, ignore_index = True)
print(f"Total Combined Rows: {len(combined_df)}")
```

```python
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
combined_df['cleaned_text'] = combined_df['Text'].str.lower()
combined_df['cleaned_text'] = combined_df['cleaned_text'].str.replace(r"http\S+|@\w+|[^\w\s]", '', regex=True)
combined_df['cleaned_text'] = [" ".join(w for w in text.split() if w not in stop_words) for text in combined_df['cleaned_text']]
```

```python
x = combined_df['cleaned_text']
y = combined_df['oh_label'].astype(int)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y , test_size = 0.2, random_state = 42)
```
